```python
import sympy as sp
```

```python
Delta_0, Delta, delta_0, delta = sp.symbols('Delta_0, Deltabar, delta_0, deltabar')
phi_0, phi, phi_f0, phi_f = sp.symbols('varphi_0, varphibar, varphi_f0, varphibar_f')
omega_r0, omega_r, omega_f0, omega_f = sp.symbols('omega_r0, omegabar_r, omega_f0, omegabar_f')
beta_0, beta = sp.symbols('beta_0, betabar')

psi_0_dot, psi_dot = sp.symbols('psidot_0, psibardot')
eps = sp.symbols('epsilon')

phi_0_dot, phi_dot, phi_f0_dot, phi_f_dot = sp.symbols('varphidot_0, varphibardot, varphidot_f0, varphibardot_f')
delta_0_dot, delta_dot = sp.symbols('deltadot_0, deltabardot')
beta_0_dot, beta_dot = sp.symbols('betadot_0, betabardot')

alpha_r0, alpha_r, alpha_f0, alpha_f = sp.symbols('alpha_r0, alphabar_r, alpha_f0, alphabar_f')

R, R_r, R_f = sp.symbols('R, R_r, R_f')
rho_r, rho_f = sp.symbols('rho_r, rho_f')
a_n, w, l_b = sp.symbols('a_n, w, l_b')
Vx0, Vx, Vy0, Vy = sp.symbols('V_x0, Vbar_x, V_y0, Vbar_y')
Vy0_dot, Vy_dot = sp.symbols('Vdot_y0, Vbardot_y0')

N_r0, N_r, N_f0, N_f = sp.symbols('N_r0, N_r, N_f0, N_f')

Y_r0, Y_r, Y_f0, Y_f = sp.symbols('Y_r0, Ybar_r, Y_f0, Ybar_f')

k_alpha_r, k_gamma_r, k_alpha_f, k_gamma_f = sp.symbols('k_alpha_r, k_gamma_r, k_alpha_f, k_gamma_f')
alpha_r0_prima, alpha_r_prima, alpha_f0_prima, alpha_f_prima = sp.symbols('alphaprime_r0, alphabarprime_r, alphaprime_f0, alphabarprime_f')
```

```python
delta_0 = (sp.cos(phi_0)*Delta_0 + sp.sin(eps)*beta_0)/sp.cos(eps)
delta = (sp.cos(phi_0)*Delta - sp.sin(phi_0)*phi*Delta_0 + sp.sin(eps)*beta)/sp.cos(eps)

psi_dot_0 = Vx0/R

phi_f0 = sp.asin(sp.sin(phi_0) + sp.cos(phi_0)*sp.sin(eps)*delta_0 + sp.cos(eps)*sp.cos(phi_0)*beta_0)
phi_f = (sp.cos(phi_0)*phi + sp.sin(eps)*(sp.cos(phi_0)*delta - sp.sin(phi_0)*phi*delta_0) + 
         sp.cos(eps)*(sp.cos(phi_0)*beta - sp.sin(phi_0)*phi*beta_0))/sp.cos(phi_f0)

omega_r0 = Vx0/(R_r + rho_r*(sp.cos(phi_0)-1))
omega_r = (Vx0*rho_r*sp.sin(phi_0)*phi)/(R_r + rho_r*(sp.cos(phi_0)-1))**2

alpha_r0 = sp.atan(Vy0/Vx0)
alpha_r = Vy/(Vx0 + Vy0*sp.tan(alpha_r0))

alpha_f0 = Delta_0 + (a_n*delta_0_dot - w*psi_0_dot - Vy0 + l_b*beta_0*sp.cos(delta_0) - rho_f*sp.cos(eps)*beta_0*sp.cos(delta_0))/Vx0
alpha_f = Delta + (a_n*delta_dot - w*psi_dot - Vy - l_b*(beta_0*sp.sin(delta_0)*delta - beta*sp.cos(delta_0)) + rho_f*(sp.cos(eps)*beta_0*sp.sin(delta_0*delta) - sp.cos(eps)*beta*sp.cos(delta_0)))/Vx0

Y_r0 = (k_alpha_r*alpha_r0_prima + k_gamma_r*phi_0)*N_r0
Y_r = (k_alpha_r*alpha_r0_prima + k_gamma_r*phi_0)*N_r + k_alpha_r*alpha_r_prima*N_r0 + k_gamma_r*phi*N_r0

Y_f0 = (k_alpha_f*alpha_f0_prima + k_gamma_f*phi_f0)*N_f0
Y_f = (k_alpha_f*alpha_f0_prima + k_gamma_f*phi_f0)*N_f + k_alpha_f*alpha_f_prima*N_f0 + k_gamma_f*phi_f*N_f0
```
